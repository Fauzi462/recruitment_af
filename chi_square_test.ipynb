```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('hasil_data.xlsx')
df.head()
```

```python
from scipy.stats import chi2_contingency
```

```python
tabel_kontingensi = pd.crosstab(df['Mengetahui Perusahaan ini Dari Mana?'], df['Competency Level'])
print(tabel_kontingensi)
print('\n' + '=' * 50 + '\n')
```

```python
chi2_stat, p_value, dof, expected = chi2_contingency(tabel_kontingensi)
```

```python
print(f'Chi-Squared : {chi2_stat:.4f}')
print(f'P-value : {p_value:.4f}')
print(f'Degree of freedom : {dof}')
print(pd.DataFrame(expected, index=tabel_kontingensi.index, columns=tabel_kontingensi.columns))
print('\n' + '=' * 50 + '\n')
```

```python
alpha = 0.05
if p_value <= alpha:
  print('Signifikan')
else:
  print('Tidak signifikan')
```
